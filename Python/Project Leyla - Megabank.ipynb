```python
# 2• Hər bir əməliyyatın müddətini əks etdirən yeni sütun yaradın (TransactionEndDatetime - TransactionStartDatetime).
```

```python
df['TransactionEndDateTime'] = pd.to_datetime(df['TransactionEndDateTime'])
df['TransactionStartDateTime'] = pd.to_datetime(df['TransactionStartDateTime'])
df['Transaction Time'] = df['TransactionEndDateTime'] - df['TransactionStartDateTime']  
df["Duration_minutes"] = df["Transaction Time"].dt.total_seconds() / 60
```

```python
#4• Ən yüksək ümumi əməliyyat məbləği olan kart sahibini müəyyənləşdirin.
```

```python
total_transactions = (df.groupby('CardholderID')['TransactionAmount'].sum().sort_values(ascending=False))

top_cardholder = total_transactions.index[0]
top_amount = total_transactions.iloc[0]

print("Ən yüksək ümumi əməliyyat məbləği olan kart sahibi:", top_cardholder)
print("Ümumi məbləğ:", top_amount)
```

```python
# 5• Qadınlar və kişilərin tranzaksiya məbləği arasında gözə çarpan ferqin olub olmamasini hipotez testi vasitəsilə araşdırın
```

```python
from scipy.stats import ttest_ind

female_transactions = df[df['Gender'] == 'F']['TransactionAmount']
male_transactions = df[df['Gender'] == 'M']['TransactionAmount']

t_stat, p_value = ttest_ind(female_transactions, male_transactions, equal_var=False) 

print("T-statistic:", t_stat)
print("P-value:", p_value)

alpha = 0.05
if p_value < alpha:
    print("Qadınlar və kişilər arasında tranzaksiya məbləğində əhəmiyyətli fərq var.")
else:
    print("Qadınlar və kişilər arasında əhəmiyyətli fərq yoxdur.")
```

```python
sns.boxplot(data=df, x='TransactionAmount')
plt.title(label='Transaction Amount uzre outliers')
plt.show()
```

```python
q1= np.quantile(df['TransactionAmount'],0.25)
q3= np.quantile(df['TransactionAmount'],0.75)
iqr= q3-q1
minn= q1 - 1.5*iqr
maxx= q3 + 1.5*iqr

plt.hist(df['TransactionAmount'])
plt.vlines(x= q1, ymin=0, ymax= 800000, color='green')
plt.vlines(x= q3, ymin=0, ymax= 800000, color='green')
plt.vlines(x= minn, ymin=0, ymax= 800000, color='red')
plt.vlines(x= maxx, ymin=0, ymax= 800000, color='red')
plt.show()
```
